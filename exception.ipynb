```python
# 사용자로부터 원 반지름 입력받은 후 둘레, 넓이 구한 후 출력하기

try:       
    number = int(input("반지름 입력 "))
    print(f"둘레 : {2 * 3.14 * number}")
    print(f"넓이 : {3.14 * number * number}")

except ValueError:
    print("입력값을 확인해주세요")
```

```python
# 오류 강제 발생 코드 : raise 오류명

class Bird:
    def fly(self):
        raise NotImplementedError

class Eagle(Bird):
    def fly(self):
        print("very fast")

eagle = Eagle()
eagle.fly()
```

```python
# 예외 클래스 생성 가능
class MyError(Exception):
    def __str__(self):
        return "허용되지 않는 별명입니다"

def say_nick(nick):
    if nick == "바보":
        raise MyError()
    print(nick)



try:
    say_nick("천사")
    say_nick("바보")


except MyError as e:
    print(e)

```

```python
# 문제 1. 랜덤 비밀번호 생성기

# 사용자로부터 생성할 비밀번호의 길이를 입력받아 랜덤 비밀번호를 생성하는 프로그램 작성
# choice()
# 비밀번호 길이는 4이상만 허용
# 숫자가 아닌 값을 입력하면 예외 처리
# from random import *
# pw = ""
# chars = "ABCDEFGHIJKLMNOPQRSTUVWXYZabcdefg~~~~~0123456789!@#$%"
# n = 0
# while n < 4:
#     try:
#         n = int(input("숫자를 입력하세요"))
#     except ValueError as e:
#         print(e)
#         break

# for i in range(0,n):
#     pw += choice(chars)


# print(pw)


from random import *

chars = "ABCDEFGHIJKLMNOPQRSTUVWXYZabcdefg~~~~~0123456789!@#$%"

while True :
    try:
        n = int(input("비밀번호 길이: "))
        if n < 4:
            print("비밀번호는 4자리 이상이어야 합니다.")
            continue
        break

    except ValueError:
        print("숫자를 입력하세요")

# pw = []
# for i in range(n):
#     pw.append(choice(chars))
pw = [choice(chars) for _ in range(n)]

print(f"랜덤 비밀번호 : {"".join(pw)}")
```

```python
# 나이 계산 프로그램
# 생년월일을 입력받아 현재 나이를 계산하는 프로그램 작성
# 입력형식 2001-08-25
# input() => '2001-08-25' => 날짜 형식
import datetime

now = datetime.datetime.now()

birth_str = input("생년월일 입력 (YYYY-MM-DD)")


# now 년도만 추출
try:
    birth = datetime.datetime.strptime(birth_str, "%Y-%m-%d")
    age = now.year - birth.year
    if(now.month, now.day) < (birth.month, birth.day):
        age -= 1
    print("나이 :", age)
except:
    print("날짜 형식이 올바르지 않습니다.")


# birth = input("생년월일을 입력하세요 ex)1111-11-11")

# print(birthday)
```

```python
# 학생 정보를 json으로 저장하기
# 학생 정보를 입력받아 json 형식으로 파일 저장
# 이름, 나이, 전공
# 읽어서 출력
import json


name = input("이름")
age = int(input("나이"))
what = input("전공")

dict1 = {'이름' : name, '나이' : age, '전공' : what}

with open("./data/myinfopro.json", "w", encoding="utf-8") as f:
    json.dump(dict1, f,ensure_ascii=False)

with open("./data/myinfopro.json", "r", encoding="utf-8") as f:
    data = json.load(f)
    print(data)
```

```python
# 다음과 같은 csv파일이 있다.
# 이름, 국어, 영어, 수학
# 홍길동,90,80,70
# 김철수,90,80,70
# 이영희,90,80,70
# 성춘향,90,80,70

# csv 파일을 읽어 학생별 평균 점수 출력
# 홍길동 : 80.0
import csv

with open("./data/average.csv", "w", encoding="utf-8", newline="") as f:
    headers = ['이름', '국어', '영어', '수학']
    writer = csv.DictWriter(f, fieldnames=headers)
    # 헤더작성
    writer.writeheader()
    writer.writerow({"이름":"홍길동","국어":88,"영어":92,"수학":70})
    writer.writerow({"이름":"김철수","국어":23,"영어":56,"수학":78})
    writer.writerow({"이름":"이영희","국어":67,"영어":34,"수학":89})
    writer.writerow({"이름":"성춘향","국어":88,"영어":22,"수학":74})


# csv.DictReader() 사용, 평균은 둘째자리까지
```

```python
with open("./data/average.csv", "r", encoding="utf-8", newline="") as f:
    reader = csv.DictReader(f)


    for row in reader:
        # print(row.get('이름'))
        avg = (int(row.get('국어')) + int(row.get('영어')) + int(row.get('수학'))) / 3
        print(f"{row.get('이름')} : {avg:.2f}") 
```
