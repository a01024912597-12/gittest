#### 정규표현식(정규식)
- 복잡한 문자열을 처리할 때 사용하는 방법
- 파이썬만의 고유한 문법이 아니라 문자열을 처리하는 모든 곳에서 사용하는 일종의 형식 언어

```python
# 주민등록번호 => 뒷자리를 모두 * 문자로 변경

# 파이썬에서 정규식 지원하는 모듈
import re

dir(re)

# re.compile() : 정규식을 작성
pattern = re.compile("ab*")
# 패턴을 특정 문자열이 포함하고 있는가?
print(pattern.match("abc"))
```

| 메서드 | 목적 |
|-----|----|
|match | 문자열의 처음부터 정규식과 매치되는지 조사|
|search | 문자열의 전체를 조사하여 정규식과 매치되는지 조사|
|findall | 정규식과 매치되는 모든 문자열(substring)을 리스트로 리턴|
|finditer | 정규식과 매치되는 모든 문자열(substring)을 이터레이터 객체로 리턴|


```python
# [문자] : [] 안에 들어있는 문자들과 매치하느냐?
p = re.compile("[abc]")

print(p.match("a"))
print(p.match("before"))
print(p.match("dude"))
```

```python
m = p.match("a")

print(f"패턴 시작 : {m.start()}")
print(f"패턴 끝 : {m.end()}")
print(f"패턴과 일치하는 문자열 반환 : {m.group()}")
print(f"일치의 시작,끝 반환 : {m.span()}")


if m:
    print(f"Match found {m.group()}")
else:
    print("No Match")
```

- 자주 사용하는 문자 표현식

| 정규표현식 | 설명 |
|-----|----|
|\d|숫자와 매치 [0-9] 와 동일한 표현식|
|\D|숫자가 아닌 것과 매치[^0-9]와 동일한 표현식|
|\s|화이트스페이스(스페이스,탭,줄바꿈 등)문자와 매치[빈칸\t\n\r\f\v] 와 동일한 표현식|
|\S|화이트스페이스(스페이스,탭,줄바꿈 등)문자가 아닌것과 매치[빈칸\t\n\r\f\v] 와 동일한 표현식|
|\w|문자 + 숫자와 매치 [a-zA-Z0-9_]와 동일한 표현식|
|\W|문자 + 숫자가 아닌 것과 매치 [a-zA-Z0-9_]와 동일한 표현식|



```python
p = re.compile("[0-9]+")
p = re.compile(r"\d+")
print(p.match("012 abc"))

# 원본 문자열에서 0~9가 아닌 문자 찾기
# p = re.compile("[^0-9]+")
p = re.compile(r"\D+")
print(p.match("012 abc"))

# 공백,탭,엔터, 문자 찾기
# p = re.compile("[ \t\n\r\f\v]+")
p = re.compile(r"\s+")
print(p.search("012 abc"))

# 공백,탭,엔터 문자가 아닌 문자 찾기
# p = re.compile("[^ \t\n\r\f\v]+")
p = re.compile(r"\S+")
print(p.search("012 abc"))

# 대소문자,숫자,_ 포함한 문자 찾기
# p = re.compile("[a-zA-Z0-9_]+")
p = re.compile(r"\w+")
print(p.search("012 abc"))

# 대소문자,숫자,_ 가 아닌 문자 찾기
# p = re.compile("[^a-zA-Z0-9_]+")
p = re.compile(r"\W+")
print(p.search("012 abc"))
```

```python
# re.I : ignorecase 

# 영어가 포함되어 있는지?

p = re.compile("[a-z]+", re.I)
print(p.match("python"))
print(p.match("Python"))
print(p.match("PYTHON"))
```

```python
# re.M : MULTILINE

data ="""python one
life is too short
python two
you need python
python three
"""

# ^ : 시작하는
# python 으로 시작하고 스페이스 문자하나,영어,숫자,_ 인 문자 찾아줘
p = re.compile(r"^python\s\w+",re.M)
print(p.findall(data))
```

```python
# | : or
p = re.compile("Crow|Servo")
m = p.match("CrowHello")
print(m)

m = p.match("ServoHello")
print(m)
```

```python
# ^ : 시작하는, $ : 끝나는
p = re.compile("^Life")
m = p.search("Life is too short")
print(m)

m = p.search("My Life")
print(m)

p = re.compile("short$")
m = p.search("Life is too short")
print(m)
```

```python
# () : 그루핑
# ABCABC 문자열이 반복되는지 조사

p = re.compile("(ABC)+")
m = p.search("ABCABCABC OK?")
print(m)
print(m.group())
print(m.group(1))
```

```python
# 이름 010-1234-5678
p = re.compile(r"(\w+)\s+(\d{3}-\d{4}-\d{4})")
m = p.search("park  010-1234-5678")
print(m)
print(m.group())
print(m.group(1))
print(m.group(2))
```

```python
# 대문자로 시작, 영어소문자 하나이상, 숫자 하나이상, 특수문자 하나이상

id = re.compile(r"(?=.*[A-Z])(?=.*[a-z])(?=.*\d)(?=.*[!@#$&])[A-Z][a-z0-9!@#$&]{5,14}")
# A12345678 -> 성공하게 되어버림
# 정규식은 패턴을 소비한다 => 이미 나온것에 대해선 기억하지 않는다. / 맞는지 아닌지만 체크함
# 그래서 정규식 앞부분에 검사하는 부분을 넣는다
print(id.fullmatch("A10101010"))
print(id.fullmatch("Aabc@1234"))
print(id.fullmatch("Aabc1234"))
print(id.fullmatch("aA@abc1234"))
```

```python
# 영문 대소문자, 숫자, 특수문자(~~~~) 포함, 8자 이상
pwd = re.compile(r"(?=.*[A-Z])(?=.*[a-z])(?=.*\d)(?=.*[!@#$&])[A-Z][a-z0-9!@#$&]{8,}$")

print(pwd.fullmatch("A1bc01010101"))
print(pwd.fullmatch("aA1bc01010101"))
print(pwd.fullmatch("A1#@bc01010101"))
print(pwd.fullmatch("A10101##0101"))

```

```python
# greedy 와 non-greedy
s = "<html><head><title>Title</title></head></html>"

# greedy
p = re.compile("<.+>")
print(p.search(s).group())

# non-greedy
# *?, {m,n}?
p = re.compile("<.+?>")
print(p.search(s).group())
```
